Below line tells Matplotlib to plot the given list of numbers.

It treats these numbers a    y-values.

plt.show()

````python
plt.plot([3,4,3.5,4,5,6])
plt.show()
````

[1,2,3,4,5] → These are your x-values.

[1,4,9,16,25] → These are your y-values (squares of x-values).

'r^-' → This is a format string that styles the plot:

'r' → Red color

'^' → Triangle-up markers for each point

'-' → Solid line connecting the points

````python
plt.grid(True)
plt.title('squares of the numbers')
plt.xlabel('The Numbers', fontsize=18, color='red')
plt.ylabel('The squares',fontsize=18, color='red')
plt.plot([1,2,3,4,5], [1,4,9,16,25], 'r^--', linewidth=3.0)
plt.show()
````

````python

import matplotlib.pyplot as plt

# Sample data
x = [1, 2, 3, 4, 5]
y = [10, 20, 25, 30, 40]

# Create scatter plot
plt.scatter(x, y, color='blue', marker='o')

# Add labels and title
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Simple Scatter Plot')

# Show the plot
plt.show()
````

```markdown
### 3D Surface Plot

This plot visualizes a 3D surface, which is useful for displaying functions of two variables (f(x, y)).
```

````python
import matplotlib.pyplot as plt
import numpy as np
from mpl_toolkits.mplot3d import Axes3D

# Generate sample data for a surface plot
# Using meshgrid to create a 2D grid of x and y values
x = np.linspace(-5, 5, 100)
y = np.linspace(-5, 5, 100)
X, Y = np.meshgrid(x, y)

# Define a function for the Z-axis (e.g., a simple paraboloid)
Z = np.sin(np.sqrt(X**2 + Y**2))

# Create a figure and a 3D axis
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

# Plot the surface
# 'cmap' sets the colormap, 'edgecolor' can add lines to the surface patches
surface = ax.plot_surface(X, Y, Z, cmap='viridis', edgecolor='none')

# Add labels and title
ax.set_xlabel('X-axis')
ax.set_ylabel('Y-axis')
ax.set_zlabel('Z-axis')
ax.set_title('Sample 3D Surface Plot')

# Add a color bar to show the mapping of colors to Z values
fig.colorbar(surface, shrink=0.5, aspect=5)

plt.show()
````
